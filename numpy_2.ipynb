# Numpy Indexing & Selection

```python
import numpy as np
arr = np.arange(0,11)
arr
```

```python
arr[8] #printing 8th index indexing start at 0 in numpy
```

```python
arr[1:5]
```

```python
arr[0:5]
```

```python
arr[0:6] 
```

```python
arr[5:]
#the 6th element in the array, but it is located at index 5
#(just like a normal Python list, indexing starts at zero in Numpy arrays). 
#Therefore, the arr[5:] notation actually grabs everything starting with the 
#index 5, not everything beyond index 5
```

```python
arr[0:5]=100
```

```python
arr
```
